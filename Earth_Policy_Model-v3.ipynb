# Balanced Framework: Sustainable Commonwealth Policy Model vs. Neoclassical MVPF Benchmark
## A Multi-Paradigm Quantitative Analysis of Public Investments and Systemic Boundaries

### Executive Overview
This interactive notebook operationalizes and contrasts two fundamentally opposing economic evaluation paradigms:
1. **The Neoclassical MVPF Framework (Marginal Value of Public Funds)**: Measures the economic return of public spending based on subjective monetary willingness-to-pay ($WTP$) and long-run net tax recoupment under existing income distributions.
2. **The Balanced Framework (Steady-State Economics) (ESJ & Thermodynamics)**: Measures objective material throughput, thermodynamic work potential (EROI, Exergy), environmental limit compliance (Carbon budgets, Aquifer recharge, Biosphere integrity), and equitable elevation above the **The Universal Baseline (Living Guarantee)** social floor.

### Core Philosophical & Institutional Conflict
- **Money as a Stand-in for Human Needs**: In neoclassical market economics, resource allocation is mediated through price signals and monetary willingness-to-pay ($WTP$). Because willingness-to-pay is bounded by ability-to-pay, a wealthy individual's preference for luxury speed (e.g., private highway express toll lanes) is quantified as having vastly greater "social welfare" than an impoverished community's life-and-death requirement for clean water, transit, or housing.
- **The Structural Distortion**: When public policy, education, and courts use money as the yardstick for performance, human dignity is forced to compete against corporate capital. Fines punish poverty, resources flow to whoever holds the most liquidity, and policies encourage infinite material extraction on a finite biophysical planet.
- **The Biophysical Alternative**: Earth System Economics reverses this relationship. Essential social foundations (The Universal Baseline) are guaranteed as non-negotiable rights, upper-bound luxury footprints are constrained through fair-share caps ($M_{\text{lim}}$), and success is measured by the thermodynamic stability of the Coupled Human-Environment System (CHANS).

## 2. Neoclassical MVPF Framework (Marginal Value of Public Funds)

### Theoretical Definition
The Marginal Value of Public Funds (MVPF) measures the ratio of the aggregate **Willingness to Pay (WTP)** of beneficiaries to the **Net Cost to Government**:

$$\text{MVPF} = \frac{\text{Total Willingness to Pay (WTP)}}{\text{Net Cost to Government}}$$

### Detailed Component Breakdown
1. **Upfront Inframarginal Cost**: Outlays paid to individuals who would have engaged in the activity anyway ($N_{\text{infra}}$):
   $$\text{Cost}_{\text{infra}} = N_{\text{infra}} \times T$$
2. **Upfront Marginal Cost**: Outlays paid to newly induced beneficiaries ($N_{\text{marg}}$):
   $$\text{Cost}_{\text{marg}} = N_{\text{marg}} \times T$$
3. **Long-Run Fiscal Recoupment**: Tax revenue returned from enhanced future economic earnings (tax rate $\tau$, earnings increase $\Delta Y$):
   $$\text{Revenue}_{\text{tax}} = (N_{\text{marg}} \times \Delta Y) \times \tau$$
4. **Net Cost to Government (Denominator)**:
   $$\text{Net Cost} = (\text{Cost}_{\text{infra}} + \text{Cost}_{\text{marg}}) - \text{Revenue}_{\text{tax}}$$
5. **Total Willingness to Pay (Numerator)**:
   $$\text{Total WTP} = \text{WTP}_{\text{infra}} + \text{WTP}_{\text{marg}} = (N_{\text{infra}} \times T) + [N_{\text{marg}} \times \Delta Y \times (1 - \tau)]$$

### The Viability Spectrum and Failure Modes of MVPF
- **$\text{MVPF} > 1.0$**: Standard cost-benefit approval. Benefits exceed public expenditure.
- **$\text{MVPF} = 1.0$**: Public expenditure break-even.
- **$0 < \text{MVPF} < 1.0$**: Inefficient policy under neoclassical terms ($WTP < \text{Net Cost}$).
- **$\text{MVPF} < 0$ (Unviable Zone)**: This occurs under two distinct toxic conditions:
  1. **Negative Willingness to Pay ($WTP < 0$)**: The public experiences net welfare destruction or active harm (e.g., pollution, health degradation, severe displacement) despite positive fiscal expenditures.
  2. **Negative Net Cost with Negative Value ($WTP < 0$ and $\text{Net Cost} < 0$)**: A predatory or extractive project that raises state tax revenue while inflicting severe net losses on public well-being.

```python
# Neoclassical Calculation Engine
def calculate_neoclassical_table(scenarios):
    records = {}
    for name, s in scenarios.items():
        wtp = s["wtp_usd"]
        net_cost = s["net_cost_usd"]
        mvpf = wtp / net_cost if net_cost != 0 else np.nan
        
        # Neoclassical evaluation logic
        if mvpf >= 1.0:
            status = "APPROVED (High Return)"
        elif mvpf > 0:
            status = "MARGINAL (Inefficient)"
        else:
            status = "UNVIABLE (Negative MVPF)"
            
        records[name] = {
            "Willingness to Pay (WTP, $)": f"${wtp:,.0f}",
            "Net Cost to Govt ($)": f"${net_cost:,.0f}",
            "Neoclassical MVPF Ratio": f"{mvpf:.2f}",
            "Neoclassical Ranking": status
        }
    return pd.DataFrame(records)

df_neoclassical = calculate_neoclassical_table(scenarios_raw)
print("=== NEOCLASSICAL MVPF EVALUATION MATRIX ===")
display(df_neoclassical)
```

## 3. Balanced Framework (Steady-State Economics) (Thermodynamics & ESJ)

### Multi-Attribute Performance Vector
Instead of collapsing human and ecological life into a single monetary scalar, the The Balanced Framework evaluates policies across a non-monetary vector of physical, ecological, and human-rights variables:

$$\mathbf{V}_{\text{Policy}} = \begin{bmatrix} \text{EROI} \\ \eta_{\text{thermo}} \\ \Delta C_{\text{budget}} \\ \text{Aquifer Recharge Rate (WFR)} \\ \text{BII} \\ \Delta\text{Universal Baseline (DLS)} \\ \text{FZR} \\ M_{\text{lim}} \\ \text{TER} \end{bmatrix}$$

### Explicit Mathematical Formulas
1. **Energy Return on Investment (EROI)**:
   $$\text{EROI} = \frac{E_{\text{delivered}}}{E_{\text{invested}}} \quad [\text{Threshold: } \text{EROI} \ge 1.0]$$
2. **Thermodynamic Exergy Efficiency ($\eta_{\text{thermo}}$)**:
   $$\eta_{\text{thermo}} = \frac{\text{Useful Exergy Output}}{\text{Primary Low-Entropy Energy Input}}$$
3. **Net Carbon Budget Impact ($\Delta C_{\text{budget}}$)**:
   $$\Delta C_{\text{budget}} = \frac{C_{\text{displaced}} - (C_{\text{construction}} + C_{\text{operational}})}{\text{Regional Transcribed Carbon Budget Cap}} \quad [\text{Threshold: } \Delta C_{\text{budget}} \ge 0.0]$$
4. **Aquifer Recharge Rate (Net Water Balance)**:
   $$\text{Aquifer Recharge Rate (WFR)} = \frac{\text{Direct + Indirect Annual Water Extraction}}{\text{Regional Aquifer Recharge Inflow Limit}} \quad [\text{Threshold: } \text{Aquifer Recharge Rate (WFR)} \le 1.0]$$
5. **Biosphere Integrity Index (BII)**:
   $$\text{BII} = \frac{\text{Intact Natural Land Area Transformed (km}^2\text{)}}{\text{Local Ecosystem Functional Integrity Threshold (km}^2\text{)}} \quad [\text{Threshold: } \text{BII} \le 0.10]$$
6. **The Universal Baseline Social Floor Delta ($\Delta\text{Universal Baseline (DLS)}$)**:
   $$\Delta\text{Universal Baseline (DLS)} = \frac{N_{\text{elevated}}}{N_{\text{underserved}}} \quad [\text{Target: Maximize social floor closure}]$$
7. **Cost per Person Elevated (Social Floor Efficiency)**:
   $$\text{Cost}_{\text{Universal Baseline (DLS)}} = \frac{\text{Net Financial Expenditure}}{N_{\text{elevated}}}$$
8. **Fair-Share Multiplier ($M_{\text{lim}}$)**:
   - Penalty factor applied to luxury overconsumption ($M_{\text{lim}} = 1.0$ for shared public goods; $M_{\text{lim}} > 1.0$ for luxury toll infrastructure).
9. **Transboundary Externality Ratio (TER)**:
   $$\text{TER} = \frac{\text{Embodied Ecological Footprint Displaced Outside Region}}{\text{Local Useful Service Energy Delivered}}$$

```python
# Earth System Biophysical Calculation Engine
def calculate_earth_system_table(scenarios, baseline=BASELINE):
    records = {}
    for name, s in scenarios.items():
        # 1. Thermodynamics
        eroi = s["E_delivered_GJ"] / s["E_invested_GJ"]
        exergy = s["thermo_efficiency"]
        
        # 2. Environmental Limits
        net_co2 = s["annual_displaced_co2"] - (s["annual_construction_co2"] + s["annual_operational_co2"])
        delta_c = net_co2 / baseline["Transcribed Carbon Budget Cap (t CO2e/yr)"]
        wfr = s["water_extraction_m3"] / baseline["Regional Aquifer Inflow / Recharge Limit (m3/yr)"]
        bii = s["land_transformed_km2"] / baseline["Biosphere Functional Integrity Threshold (km2 grid)"]
        
        # 3. Social Floor & Friction
        delta_dls = s["elevated_dls_pop"] / baseline["Underserved Population Below Universal Baseline (DLS) (N_underserved)"]
        cost_dls = s["net_cost_usd"] / s["elevated_dls_pop"]
        fzr = s["roadblock_removal_score"]
        
        # 4. Justice & Transboundary
        m_lim = s["fair_share_multiplier"]
        ter = s["out_of_region_footprint_GJ"] / s["E_delivered_GJ"]
        
        # 5. Boundary Compliance Checks
        c_pass = delta_c >= 0.0
        w_pass = wfr <= 1.0
        b_pass = bii <= 0.10
        s_pass = delta_dls >= 0.50
        overall_pass = c_pass and w_pass and b_pass and s_pass
        
        records[name] = {
            "EROI (Net Energy Ratio)": f"{eroi:.2f}",
            "Exergy Efficiency (eta)": f"{exergy:.2f}",
            "Net CO2 Impact (Delta C_budget)": f"{delta_c:+.2f}",
            "Aquifer Recharge Rate (Net Water Balance)": f"{wfr:.3f}",
            "Biosphere Integrity (BII)": f"{bii:.3f}",
            "Universal Baseline (DLS) Social Floor Gain (Delta Universal Baseline (DLS))": f"{delta_dls*100:.1f}%",
            "Cost per Universal Baseline (DLS) Person Elevated": f"${cost_dls:,.0f}",
            "Friction Zone Reduction (FZR)": f"{fzr}/100",
            "Fair-Share Reallocation (M_lim)": f"{m_lim}x",
            "Transboundary Externality (TER)": f"{ter:.2f}",
            "Biophysical Boundary Verdict": "PASS (Within Limits)" if (c_pass and w_pass and b_pass) else "FAIL (Overshoot)",
            "Earth System Overall Verdict": "APPROVED" if overall_pass else "REJECTED"
        }
    return pd.DataFrame(records)

df_earth_system = calculate_earth_system_table(scenarios_raw)
print("=== EARTH SYSTEM BIOPHYSICAL EVALUATION MATRIX ===")
display(df_earth_system)
```

## 4. Constraint-Bounded Integration: Unconstrained MVPF vs. Safe MVPF

### The Paradigm Divergence
When we compare Neoclassical MVPF directly against the The Balanced Framework, a profound paradox emerges:
- **Option B (Suburban Highway Expansion)** scores the highest Neoclassical MVPF ($\mathbf{4.20}$) because affluent suburban commuters have a high willingness-to-pay for private speed, generating substantial toll and fuel revenues.
- However, under Earth System Biophysical Accounting, Option B is **categorically unviable**: it severely breaches the commonwealth district carbon budget ($\Delta C_{\text{budget}} = -0.65$), fragments ecosystems ($\text{BII} = 0.45$), triggers a $2.5\times$ limitarian luxury penalty, and leaves $36,000$ vulnerable residents trapped below the Universal Baseline (DLS) mobility floor.

### The Constraint-Bounded Filter (Safe MVPF)
To reconcile these paradigms, we apply the **Earth System Constraint Filter**:

$$\text{Safe MVPF} = \begin{cases} \text{Raw MVPF}, & \text{if } \Delta C_{\text{budget}} \ge 0 \;\wedge\; \text{Aquifer Recharge Rate (WFR)} \le 1.0 \;\wedge\; \text{BII} \le 0.10 \;\wedge\; \Delta\text{Universal Baseline (DLS)} \ge 0.50 \\ 0.0, & \text{if any biophysical ceiling or social floor is violated} \end{cases}$$

A policy that destroys the biophysical stability of the planet or deepens social deprivation is filtered to zero social return, regardless of how much private monetary willingness-to-pay it commands.

```python
# Side-by-Side Paradigm Comparison Table
def calculate_side_by_side(scenarios, baseline=BASELINE):
    comparison = {}
    for name, s in scenarios.items():
        # Neoclassical
        raw_mvpf = s["wtp_usd"] / s["net_cost_usd"]
        
        # Biophysical
        net_co2 = s["annual_displaced_co2"] - (s["annual_construction_co2"] + s["annual_operational_co2"])
        delta_c = net_co2 / baseline["Transcribed Carbon Budget Cap (t CO2e/yr)"]
        wfr = s["water_extraction_m3"] / baseline["Regional Aquifer Inflow / Recharge Limit (m3/yr)"]
        bii = s["land_transformed_km2"] / baseline["Biosphere Functional Integrity Threshold (km2 grid)"]
        delta_dls = s["elevated_dls_pop"] / baseline["Underserved Population Below Universal Baseline (DLS) (N_underserved)"]
        
        is_safe = (delta_c >= 0.0) and (wfr <= 1.0) and (bii <= 0.10) and (delta_dls >= 0.50)
        safe_mvpf = raw_mvpf if is_safe else 0.0
        
        comparison[name] = {
            "Neoclassical MVPF (WTP/Cost)": f"{raw_mvpf:.2f}",
            "Constraint-Bounded Safe MVPF": f"{safe_mvpf:.2f}",
            "Carbon Budget Headroom": f"{delta_c:+.2f}",
            "Universal Baseline (DLS) Floor Coverage": f"{delta_dls*100:.1f}%",
            "Fair-Share Reallocation Multiplier": f"{s['fair_share_multiplier']}x",
            "Paradigm Recommendation": "APPROVED (Pareto Dominant)" if is_safe else "REJECTED (Ecological Overshoot)"
        }
    return pd.DataFrame(comparison)

df_comparison = calculate_side_by_side(scenarios_raw)
print("=== SIDE-BY-SIDE PARADIGM COMPARISON ===")
display(df_comparison)
```

## 5. Visualizing the Paradox & The Unviable Negative MVPF Zone

Below, we visualize two critical structural dynamics:
1. **Figure 1: Paradigm Inversion** — Side-by-side bar chart showing how Option B collapses from a neoclassical score of 4.20 to an Earth System score of 0.00.
2. **Figure 2: The MVPF Viability Spectrum** — A continuous function of Willingness-to-Pay ($WTP$) versus Net Cost, explicitly mapping the **Negative MVPF Unviable Zone ($MVPF < 0$)**.

```python
# Plot 1: Paradigm Divergence Bar Chart
fig, ax = plt.subplots(figsize=(11, 5.5))
options = list(scenarios_raw.keys())
x = np.arange(len(options))
width = 0.35

raw_vals = [s["wtp_usd"] / s["net_cost_usd"] for s in scenarios_raw.values()]
safe_vals = []
for s in scenarios_raw.values():
    net_co2 = s["annual_displaced_co2"] - (s["annual_construction_co2"] + s["annual_operational_co2"])
    delta_c = net_co2 / BASELINE["Transcribed Carbon Budget Cap (t CO2e/yr)"]
    bii = s["land_transformed_km2"] / BASELINE["Biosphere Functional Integrity Threshold (km2 grid)"]
    dls = s["elevated_dls_pop"] / BASELINE["Underserved Population Below Universal Baseline (DLS) (N_underserved)"]
    if delta_c >= 0 and bii <= 0.10 and dls >= 0.50:
        safe_vals.append(s["wtp_usd"] / s["net_cost_usd"])
    else:
        safe_vals.append(0.0)

rects1 = ax.bar(x - width/2, raw_vals, width, label='Unconstrained Neoclassical MVPF (WTP/Net Cost)', color='#3498db', edgecolor='black')
rects2 = ax.bar(x + width/2, safe_vals, width, label='Balanced Framework Bounded Safe MVPF (Boundary Filtered)', color='#2ecc71', edgecolor='black')

ax.set_ylabel('MVPF Ratio (Benefit per $1 of Net Cost)', fontsize=12, fontweight='bold')
ax.set_title('Paradigm Inversion: Unconstrained Monetary Optimization vs. Biophysical Boundary Filtering', fontsize=13, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(options, fontsize=11)
ax.axhline(1.0, color='red', linestyle='--', linewidth=1.5, label='Neoclassical Break-even Line (1.0)')
ax.legend(fontsize=11, loc='upper right')

# Annotate bar values
for r in rects1:
    h = r.get_height()
    ax.annotate(f'{h:.2f}', xy=(r.get_x() + r.get_width()/2, h), xytext=(0, 4),
                textcoords="offset points", ha='center', va='bottom', fontweight='bold')
for r in rects2:
    h = r.get_height()
    ax.annotate(f'{h:.2f}', xy=(r.get_x() + r.get_width()/2, h), xytext=(0, 4),
                textcoords="offset points", ha='center', va='bottom', fontweight='bold', color='#1e8449' if h>0 else 'red')

plt.tight_layout()
plt.show()
```

```python
def live_policy_simulation(delivered_gj, invested_gj, construction_co2, op_co2, displaced_co2, water_m3, land_km2, dls_elevated, wtp_usd):
    net_co2 = displaced_co2 - (construction_co2 + op_co2)
    delta_c = net_co2 / BASELINE["Transcribed Carbon Budget Cap (t CO2e/yr)"]
    wfr = water_m3 / BASELINE["Regional Aquifer Inflow / Recharge Limit (m3/yr)"]
    bii = land_km2 / BASELINE["Biosphere Functional Integrity Threshold (km2 grid)"]
    delta_dls = dls_elevated / BASELINE["Underserved Population Below Universal Baseline (DLS) (N_underserved)"]
    eroi = delivered_gj / invested_gj
    
    # Fiscal
    net_cost = 10000000
    raw_mvpf = wtp_usd / net_cost
    
    # Boundary tests
    c_pass = delta_c >= 0.0
    w_pass = wfr <= 1.0
    b_pass = bii <= 0.10
    s_pass = delta_dls >= 0.50
    is_safe = c_pass and w_pass and b_pass and s_pass
    safe_mvpf = raw_mvpf if is_safe else 0.0
    
    sim_df = pd.DataFrame({
        "Metric": [
            "Energy Return on Investment (EROI)",
            "Net Annual CO2 Balance (t CO2e)",
            "Carbon Budget Impact (Delta C_budget)",
            "Aquifer Recharge Rate (Net Water Balance)",
            "Biosphere Integrity Index (BII)",
            "Universal Baseline (DLS) Social Floor Gain (%)",
            "Cost per Universal Baseline (DLS) Person Elevated ($)",
            "Unconstrained Neoclassical MVPF",
            "Balanced Framework Bounded Safe MVPF",
            "Overall Policy Verdict"
        ],
        "Simulated Value": [
            f"{eroi:.2f}",
            f"{net_co2:+,.0f} t",
            f"{delta_c:+.2f}",
            f"{wfr:.3f}",
            f"{bii:.3f}",
            f"{delta_dls*100:.1f}%",
            f"${net_cost/dls_elevated:,.0f}",
            f"{raw_mvpf:.2f}",
            f"{safe_mvpf:.2f}",
            "APPROVED (Safe & Just)" if is_safe else "REJECTED (Boundary Violation)"
        ],
        "Threshold Rule": [
            ">= 1.0 (Thermodynamic Breakeven)",
            ">= 0 (Emissions Reduction)",
            ">= 0.0 (Budget Headroom)",
            "<= 1.0 (Aquifer Recharge Limit)",
            "<= 0.10 (Local Habitat Integrity)",
            ">= 50% (Substantial Floor Closure)",
            "Minimize Cost per Beneficiary",
            ">= 1.0 (Neoclassical Breakeven)",
            ">= 1.0 with Safe Boundaries",
            "Simultaneous Boundary Compliance"
        ]
    })
    display(sim_df)

# Display Interactive Widget Controls
interact(live_policy_simulation,
         delivered_gj=widgets.IntSlider(min=50000, max=200000, step=5000, value=110000, description='Delivered GJ:'),
         invested_gj=widgets.IntSlider(min=10000, max=60000, step=2000, value=22000, description='Invested GJ:'),
         construction_co2=widgets.IntSlider(min=1000, max=30000, step=1000, value=6000, description='Const CO2:'),
         op_co2=widgets.IntSlider(min=5000, max=70000, step=2000, value=12000, description='Op CO2:'),
         displaced_co2=widgets.IntSlider(min=10000, max=60000, step=2000, value=40000, description='Displaced CO2:'),
         water_m3=widgets.IntSlider(min=50000, max=1000000, step=25000, value=180000, description='Water (m3):'),
         land_km2=widgets.FloatSlider(min=0.1, max=5.0, step=0.1, value=0.8, description='Land (km2):'),
         dls_elevated=widgets.IntSlider(min=1000, max=40000, step=1000, value=28000, description='Universal Baseline (DLS) Elevated:'),
         wtp_usd=widgets.IntSlider(min=-20000000, max=50000000, step=1000000, value=21000000, description='Public WTP ($):'));
```
